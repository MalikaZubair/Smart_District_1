```python
#Data + ranking
employees = [
    ("Ayesha Khan",     92, 5),    # (name, score, experience_years)
    ("Bilal Ahmed",     88, 7),
    ("Chaudhry Usman",  92, 8),
    ("Danish Ali",      88, 7),
    ("Fatima Noor",     95, 3),
    ("Hamza Sheikh",    92, 5),
    ("Iqra Malik",      75, 2),
    ("Zainab Raza",     88, 4),
]

ranked = [
    {"rank": i, "name": n, "score": s, "experience": x}
    for i, (n, s, x) in enumerate(
        sorted(employees, key=lambda e: (-e[1], -e[2], e[0].lower())),
        start=1,
    )
]

print(f"{'Rank':<6}{'Name':<18}{'Score':>6}{'Exp':>5}")
print("-" * 35)
print("\n".join(
    f"{e['rank']:<6}{e['name']:<18}{e['score']:>6}{e['experience']:>5}" for e in ranked
))
```

```python
#Fast search (binary search)
from bisect import bisect_left, bisect_right

by_name = sorted(ranked, key=lambda e: e["name"].lower())
name_keys = [e["name"].lower() for e in by_name]

def find_employee(name):
    """Exact naam se dhoondo. List return hoti hai (khali = nahi mila)."""
    q = name.strip().lower()
    return by_name[bisect_left(name_keys, q):bisect_right(name_keys, q)]

def search_prefix(prefix):
    """Naam ki shuruaat se dhoondo, e.g. 'ha' -> Hamza Sheikh."""
    q = prefix.strip().lower()
    return by_name[bisect_left(name_keys, q):bisect_left(name_keys, q + "\uffff")]
```

```python
#Search test
for q in ["bilal ahmed", "  HAMZA SHEIKH ", "Nobody"]:
    hits = find_employee(q)
    if hits:
        e = hits[0]
        print(f"✅ {e['name']}: rank #{e['rank']}, score {e['score']}, exp {e['experience']} yrs")
    else:
        print(f"❌ '{q.strip()}' nahi mila")

print("\nPrefix 'a':", [e["name"] for e in search_prefix("a")])
```
